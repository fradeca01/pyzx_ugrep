```python
import stim
import pyzx
import matplotlib.pyplot as plt
import networkx as nx
from ugr import *
```

```python
n = 9
k = 1
shor_stabilizers = [stim.PauliString("+ZZ_______"), stim.PauliString("+_ZZ______"), stim.PauliString("+___ZZ____"), stim.PauliString("+____ZZ___"), stim.PauliString("+______ZZ_"), stim.PauliString("+_______ZZ"), stim.PauliString("+XXXXXX___"), stim.PauliString("+___XXXXXX"), stim.PauliString("+__Z__Z__Z")]
```

```python
# u = stabilizers_to_UGR(shor_code["stabilizers"])
tableau = stim.Tableau.from_stabilizers(shor_stabilizers)


encoder_circuit = tableau.to_circuit(method = "elimination")

print(encoder_circuit.diagram())

qasm = encoder_circuit.to_qasm(open_qasm_version=3)



pyzx_circ = pyzx.circuit.Circuit.from_qasm(qasm)
pyzx.draw(pyzx_circ)

input_state = "0"*(n - k) + "/"*k
g2 = pyzx_circ.to_graph()
g2.apply_state(input_state)

pyzx.draw(g2)
```

```python
from qiskit import qasm3
from qiskit.visualization import circuit_drawer

qasm = encoder_circuit.to_qasm(open_qasm_version=3)

qc = qasm3.loads(qasm)

latex_source = circuit_drawer(qc, output="latex_source", fold=-1)

with open("encoder.tex", "w") as f:
    f.write(latex_source)
```
